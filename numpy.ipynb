```python
import numpy as np
import pandas as pd
car=np.array([1,2,3,4,5])
print(np.max(car))
```

```python
np1=np.array([1,2,3,4,5])
print(np.sin(np1))
print(np.log(np1))
print(np.sqrt(np1[3]))
```

```python
np3=np.array([1,3,4,-3,4])
np4=np3.copy()
print(f'original value{np3}')
print(f'changed value{np4}')
np4[3]=27
print(f'original value{np3}')
print(f'changed value{np4}')
```
